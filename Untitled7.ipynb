```python
#-Declaracion de variables
arreglo_1= [1,2,3,4,5]
arreglo_2= ["juan","leo","gabriel"]
arreglo_3= ["juan",13,"estudiante"]

#-Salidas
print(arreglo_1)
print(arreglo_2)
print(arreglo_3)
```

```python
#-Seleccion
componente_1=arreglo_1[0]
componente_2=arreglo_2[0]
componente_3=arreglo_3[0]

#---------Salida----------------------
print(componente_1)
print(componente_2)
print(componente_3)
```

```python
#-Modificar componente-
arreglo_1[0]= 10
arreglo_2[0]="juan"
arreglo_3[0]="david"

#Salida
print(arreglo_1)
print(arreglo_2)
print(arreglo_3)

```

```python
#-Ingresar nuevo componente
nuevo_componente=100

arreglo_1.append(nuevo_componente)
arreglo_2.append(nuevo_componente)
arreglo_3.append(nuevo_componente)

#---------Salida----------------------
print(arreglo_1)
print(arreglo_2)
print(arreglo_3)
```

```python
arreglo_1.remove(100)
arreglo_2.remove("juan")
arreglo_3.remove(arreglo_3[0])

#---------Salida----------------------
print(arreglo_1)
print(arreglo_2)
print(arreglo_3)
```

```python
#-Estructura de contro para recorrer
for componente_actual in arreglo_1:
  print(componente_actual)
```

```python
#Estructura de control para recorrer
for componente_actual in arreglo_2:
  print(componente_actual)
```

```python
#Estructura de control para recorrer
for componente_actual in arreglo_3:
  print(componente_actual)
```

```python
#-Estructura de contro para recorrer
for componente_actual in range(5):
  print(componente_actual)
```

```python
#-Estructura de contro para recorrer
for componente_actual in range(len(arreglo_1)):
  print(componente_actual)
```
